```python
from smolagents import CodeAgent, DuckDuckGoSearchTool, InferenceClientModel
```

```python

# 1. Instantiate the DuckDuckGoSearchTool
web_search = DuckDuckGoSearchTool()

# 2. Choose and instantiate your desired language model
#    Here, we use HfApiModel, but you can use others like InferenceClientModel
#    Replace "your_model_id" with the actual ID of the Hugging Face model you want to use
# model = HfApiModel(model_id="your_model_id") 
model = InferenceClientModel(
    max_tokens=2096,
    temperature=0.5,
    model_id='Qwen/Qwen2.5-Coder-32B-Instruct',
    custom_role_conversions=None,
)


# 3. Create an agent (e.g., CodeAgent or ToolCallingAgent) and pass the search tool
agent = CodeAgent(tools=[web_search], model=model)

# 4. Run the agent with a prompt that requires a web search
query = "Who won the St. Jude's classic golf tournament in 2025?"
result = agent.run(query)

# 5. Print the result
print(result)
```
