```python
from smolagents import CodeAgent, DuckDuckGoSearchTool, InferenceClientModel

# 1. Instantiate the DuckDuckGoSearchTool
web_search = DuckDuckGoSearchTool()

# 2. Choose and instantiate your desired language model
#    Here, we use HfApiModel, but you can use others like InferenceClientModel
#    Replace "your_model_id" with the actual ID of the Hugging Face model you want to use
# model = HfApiModel(model_id="your_model_id") 
model = InferenceClientModel(
    max_tokens=2096,
    temperature=0.5,
    model_id='Qwen/Qwen2.5-Coder-32B-Instruct',
    custom_role_conversions=None,
)


# 3. Create an agent (e.g., CodeAgent or ToolCallingAgent) and pass the search tool
agent = CodeAgent(tools=[web_search], model=model)

# 4. Run the agent with a prompt that requires a web search
query = "Who won the St. Jude's classic golf tournament in 2025?"
result = agent.run(query)

# 5. Print the result
print(result)
```

```python
import os
from huggingface_hub import InferenceClient

## You need a token from https://hf.co/settings/tokens, ensure that you select 'read' as the token type. If you run this on Google Colab, you can set it up in the "settings" tab under "secrets". Make sure to call it "HF_TOKEN"
HF_TOKEN = os.environ.get("HUGGINGFACE_API_KEY")

client = InferenceClient(model="meta-llama/Llama-4-Scout-17B-16E-Instruct", token=HF_TOKEN)
```

```python
output = client.chat.completions.create(
    messages=[
        {"role": "user", "content": "The capital of France is"},
    ],
    stream=False,
    max_tokens=20,
)
print(output.choices[0].message.content)
```

```python
from smolagents import CodeAgent, InferenceClientModel

# Initialize a model (using Hugging Face Inference API)
# model = InferenceClientModel()  # Uses a default model
model = InferenceClientModel(token=HF_TOKEN)

# Create an agent with no tools
agent = CodeAgent(tools=[], model=model)

# Run the agent with a task
result = agent.run("Calculate the sum of numbers from 1 to 10")
print(result)
```
